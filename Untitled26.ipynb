```python
import random

# Generate random number between 1 and 10
secret_number = random.randint(1, 10)

attempts = 3

while attempts > 0:
    guess = int(input("Guess the number (between 1 and 10): "))

    # Check if guess is out of range
    if guess < 1 or guess > 10:
        print("Your guess is out of range. Please guess a number between 1 and 10.")
        continue  # Skip the rest of the loop, don't count this attempt

    attempts -= 1  # Only count valid guesses

    if guess == secret_number:
        print("Congratulations! You guessed the correct number.")
        break  # Exit the loop when correct
    elif guess > secret_number:
        print("Too high. Try again.")
    else:
        print("Too low. Try again.")
else:
    # This executes only if the loop finishes without a break
    print(f"Better luck next time! The number was {secret_number}.")
```

```python

#2. Multiplication Table Generator (For Loop)
# Prompt user for input
number = int(input("Enter the number for which you want the multiplication table: "))

# Generate and display the table
for i in range(1, 12):
    result = number * i
    print(f"{number} x {i} = {result}")
```

```python

#3. BMI Calculator (Function)
def calculate_bmi(weight, height):
    """Calculate BMI given weight in kg and height in meters."""
    bmi = weight / (height ** 2)
    return bmi

# Prompt user for weight and height
weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

# Calculate and display BMI
bmi = calculate_bmi(weight, height)
print(f"Your BMI is: {bmi:.2f}")
```
