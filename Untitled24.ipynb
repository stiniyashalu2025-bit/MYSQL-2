```python
# 1. List Creation
age_list = [24, 25, 26, 27, 28]
name_list = ["Stiniya", "Shalu", "Stinith","Jobin", "Joel"]

# 2. List Operations / Modifications
# a. Append the string "Yazhini" to name_list
name_list.append("Yazhini")

# b. Insert the element 30 at index 2 in age_list
age_list.insert(2, 30)

# c. Remove the string "Yazhini" from name_list
name_list.remove("Yazhini")

# d. Pop the last element from age_list
age_list.pop()

# e. Extend the age_list with additional ages [29, 30, 26]
age_list.extend([29, 30, 26])

# f. Sort age_list in descending order
age_list.sort(reverse=True)

# g. Find Max age, Min age and sum of all ages from age_list
max_age = max(age_list)
min_age = min(age_list)
sum_ages = sum(age_list)
print(f"Max: {max_age}, Min: {min_age}, Sum: {sum_ages}")

# 3. Accessing List Elements
# a. Print the first element of name_list
print(name_list[0])

# b. Print the last element of name_list
print(name_list[-1])

# c. Print the elements from index 2 to index 4 in name_list
print(name_list[2:5]) # Slicing is exclusive of the end index

# d. Print the elements of name_list in reverse order
print(name_list[::-1])
```

```python
# Performance Category Program

# 1. Prompt user for input
try:
    score = float(input("Enter your score (0 to 10): "))

    # Basic validation to ensure score is in range
    if score < 0 or score > 10:
        print("Invalid input. Score must be between 0 and 10.")
    else:
        # 2. Find the performance category
        if score > 7:
            # a. Above Average
            category = "Above Average"
            prompt = "Excellent work! Keep it up."
        elif 4 <= score <= 7:
            # b. Average
            category = "Average"
            prompt = "Good effort! Keep practicing, there's room for improvement."
        else:
            # c. Below Average (score < 4)
            category = "Below Average"
            prompt = "Need to improve your performance, consistent practice will lead to better results."

        # 3. Output: Print the Performance category
        # 4. Additional Step: Print the prompt
        print(f"{category}: {prompt}")

except ValueError:
    print("Invalid input. Please enter a numeric value.")
```
