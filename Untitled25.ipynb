```python
#1. Create a 1D NumPy Array:
import pandas as pd
import numpy as np
temperatures_w1 = np.array([12.5, 22.3, 32.8, 42.4, 52.1, 62.8, 72.9])
print(temperatures_w1)
```

```python
#2. Inspection and Properties:
print(temperatures_w1.shape)
print(temperatures_w1.dtype)
print(temperatures_w1.size)
```

```python
#3. Array Operations:
temperatures_f = (temperatures_w1 * 8/6) + 22
print(temperatures_f)

print(np.max(temperatures_w1))
print(np.min(temperatures_w1))
print(np.mean(temperatures_w1))
```

```python
#4. Array Slicing and Indexing:
print(temperatures_w1[:3])
print(temperatures_w1[-2:])
print(temperatures_w1[2:5])
```

```python
#5. Create a 2D Array:
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])
print(temperatures)
```

```python
##6. Inspect and Slice the 2D Array:
print(temperatures.shape)
print(temperatures.dtype)
print(temperatures.size)

print(temperatures[:, :2])
print(temperatures[:, -2:])
```

```python
##PANDAS SERIES
#1. Creating Pandas Series:
import pandas as pd
marks = pd.Series([95, 92, 89, 85, 80], index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print(marks)
```

```python
#2. Indexing and Slicing:
print(marks.iloc[0])
print(marks.loc['Rank1':'Rank3'])
print(marks.iloc[2])
print(marks[marks > 90])
```

```python
#3. Manipulating Series:
marks['Rank1'] = 100
print(marks)

marks = marks.drop('Rank1')
print(marks)

cgpa = marks / 10
print(cgpa)
```

```python
##PANDAS DATAFRAME
#1. Creating Pandas DataFrame:
data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}
transactions = pd.DataFrame(data)
(transactions)
```

```python
#2. Data Exploration:
(transactions.head())
(transactions.tail())
(transactions.shape)
(transactions.columns.tolist())
(transactions.dtypes)

(transactions[['ProductCategory', 'Amount']])
(transactions.iloc[:, -3:])
(transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)])
(transactions['ProductCategory'].value_counts())
(transactions['Region'].unique())
(transactions.groupby('Region')['Amount'].mean())
```

```python
#3. Manipulating the DataFrame:
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
(transactions)

transactions['Discount'] = transactions['Amount'] * 0.10
(transactions)

transactions = transactions[transactions['TransactionID'] != 109]
(transactions)

transactions = transactions.drop(columns=['Discount'])
(transactions)
```
